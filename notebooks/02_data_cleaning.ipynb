# 02 - Data preparation

Objective: standardize, validate, and combine yearly NSE files without modifying the raw downloads. Only the `EQ` series is retained in the processed dataset.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
```

```python
from finsight.data import build_processed_india, load_processed_india, INDIAN_STOCKS

raw = ROOT / "data" / "raw" / "india" / "stocks"
processed = ROOT / "data" / "processed" / "india" / "eq_daily.parquet"
if processed.exists():
    print("Processed output already exists; set overwrite=True only when intentionally rebuilding.")
else:
    build_processed_india(raw, processed)
selected = load_processed_india(processed, symbols=INDIAN_STOCKS)
display(selected.head())
print(f"Selected stock rows: {len(selected):,}; symbols: {selected.symbol.nunique()}")
```

Interpretation: validation raises an explicit error for null required fields, nonpositive prices, invalid OHLC relationships, negative volume, or duplicate keys; it never silently drops such rows.

Conclusion: the full `EQ` Parquet is ready for downstream analysis. Corporate-action adjustment remains unresolved.
