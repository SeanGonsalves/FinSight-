# 03 - Exploratory analysis

Objective: summarize the selected Indian equities, their price/return distributions, volume, and data coverage.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
```

```python
from finsight.data import load_processed_india, INDIAN_STOCKS
from finsight.visualization import plot_price_distributions, plot_return_distributions, plot_volume

data = load_processed_india(ROOT / "data" / "processed" / "india" / "eq_daily.parquet", symbols=INDIAN_STOCKS)
data = data.loc[data.date.between("2016-01-01", "2025-12-31")].copy()
print(data.date.min(), data.date.max(), len(data), data.symbol.nunique())
display(data.groupby("symbol")["close"].describe())
display(data.groupby("symbol")["volume"].describe())
figures = ROOT / "reports" / "figures" / "eda"
plot_price_distributions(data, INDIAN_STOCKS, figures / "closing_price_distributions.png")
plot_return_distributions(data, INDIAN_STOCKS, figures / "daily_return_distributions.png")
plot_volume(data, INDIAN_STOCKS, figures / "trading_volume.png")
missing = data.set_index("symbol").isna().groupby(level="symbol").sum().sum(axis=1)
display(missing.rename("missing_cells"))
```

Interpretation: compare trading activity and distributions across instruments, while remembering that volume units and price adjustments are not documented by the source.

Conclusion: record market-wide findings only from observed data; the audited yearly file date counts are in the dataset audit report.
