# 04 - Stock performance

Objective: compute simple daily, monthly, and annual returns and compare normalized closes for the available Indian stock set.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
```

```python
from finsight.data import load_processed_india, INDIAN_STOCKS
from finsight.metrics import daily_returns, periodic_returns, annual_returns
from finsight.visualization import plot_monthly_returns, plot_normalized_prices

data = load_processed_india(ROOT / "data" / "processed" / "india" / "eq_daily.parquet", symbols=INDIAN_STOCKS)
data = data.loc[data.date.between("2016-01-01", "2025-12-31")].copy()
returns = daily_returns(data)
monthly = periodic_returns(data, "M")
yearly = annual_returns(data)
display(monthly.head(10))
display(yearly)
plot_normalized_prices(data, INDIAN_STOCKS, ROOT / "reports" / "figures" / "performance" / "normalized_closes.png")
plot_monthly_returns(monthly, ROOT / "reports" / "figures" / "performance" / "monthly_returns.png")
```

Interpretation: normalized closing-price comparisons are descriptive and may be distorted by undocumented split/dividend adjustment. The returns are not total returns.
