# 05 - Risk and volatility

Objective: compare annualized volatility, maximum drawdown, Sharpe ratio, and downside deviation using a consistent 252-session convention.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
```

```python
from finsight.data import load_processed_india, INDIAN_STOCKS
from finsight.metrics import risk_performance_summary
from finsight.visualization import plot_rolling_volatility, plot_drawdown

data = load_processed_india(ROOT / "data" / "processed" / "india" / "eq_daily.parquet", symbols=INDIAN_STOCKS)
data = data.loc[data.date.between("2016-01-01", "2025-12-31")].copy()
risk = risk_performance_summary(data, risk_free_rate=0.0, periods_per_year=252)
display(risk.sort_values("annualized_volatility", ascending=False))
figures = ROOT / "reports" / "figures" / "risk"
plot_rolling_volatility(data, INDIAN_STOCKS, figures / "rolling_volatility.png")
plot_drawdown(data, INDIAN_STOCKS, figures / "drawdowns.png")
```

Interpretation: the Sharpe ratio assumes a zero risk-free rate; drawdown and volatility are estimated from observed source closes. No measure is investment advice.

Conclusion: compare like-for-like coverage and price basis before drawing risk rankings.
