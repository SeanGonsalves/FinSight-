# 06 - Correlation and statistical analysis

Objective: measure Pearson and Spearman dependence in daily returns, test one selected pair, and run ADF tests on a price and return series.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
```

```python
from finsight.data import load_processed_india, INDIAN_STOCKS
from finsight.metrics import correlation_matrix, correlation_significance, rolling_correlation, adf_stationarity_test
from finsight.visualization import plot_correlation_heatmap, plot_rolling_correlation

data = load_processed_india(ROOT / "data" / "processed" / "india" / "eq_daily.parquet", symbols=INDIAN_STOCKS)
data = data.loc[data.date.between("2016-01-01", "2025-12-31")].copy()
pearson = correlation_matrix(data, method="pearson")
spearman = correlation_matrix(data, method="spearman")
rolling = rolling_correlation(data, "RELIANCE", "TCS", window=60)
display(pearson)
display(spearman)
print(correlation_significance(data, "RELIANCE", "TCS"))
stock = data.loc[data.symbol.eq("RELIANCE")].sort_values("date")
returns = stock.close.pct_change(fill_method=None)
print("Close ADF:", adf_stationarity_test(stock.close))
print("Return ADF:", adf_stationarity_test(returns))
plot_correlation_heatmap(pearson, ROOT / "reports" / "figures" / "correlation" / "pearson_heatmap.png")
plot_rolling_correlation(rolling, "RELIANCE", "TCS", ROOT / "reports" / "figures" / "correlation" / "reliance_tcs_rolling_correlation.png")
```

Interpretation: p-values depend on test assumptions and sample selection; correlation does not imply causation. ADF results do not prove an economic mechanism.
