# 07 - Technical indicators

Objective: compute SMA, EMA, RSI, MACD, and Bollinger Bands for a selected stock using only current and past observations.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
```

```python
from finsight.data import load_processed_india
from finsight.indicators import add_technical_indicators
from finsight.visualization import plot_technical_indicators

data = load_processed_india(ROOT / "data" / "processed" / "india" / "eq_daily.parquet", symbols=["RELIANCE"])
data = data.loc[data.date.between("2016-01-01", "2025-12-31")].copy()
indicators = add_technical_indicators(data)
display(indicators.tail())
plot_technical_indicators(indicators, "RELIANCE", ROOT / "reports" / "figures" / "technical" / "reliance_indicators.png")
```

Interpretation: technical indicators describe transformations of historical prices; they are not evidence of profitable predictive signals. Warm-up periods naturally produce missing indicator values.

Conclusion: corporate-action adjustment remains important for interpreting long histories.
