# 08 - Predictive modelling

Objective: compare a zero-return baseline and regression models for next-observed-session returns using chronological train/validation/test splits.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
```

```python
from finsight.data import load_processed_india
from finsight.modeling import evaluate_forecasts

stock = load_processed_india(ROOT / "data" / "processed" / "india" / "eq_daily.parquet", symbols=["RELIANCE"])
stock = stock.loc[stock.date.between("2016-01-01", "2025-12-31")].copy()
forecast = evaluate_forecasts(stock)
print("Split sizes:", forecast.split_sizes)
print("Validation-selected model:", forecast.selected_model)
display(forecast.validation_metrics)
outputs = ROOT / "data" / "processed" / "combined"
outputs.mkdir(parents=True, exist_ok=True)
forecast.predictions.to_parquet(outputs / "forecast_predictions.parquet", index=False)
tables = ROOT / "reports" / "tables"
tables.mkdir(parents=True, exist_ok=True)
forecast.validation_metrics.to_csv(tables / "model_validation.csv", index=False)
forecast.metrics.to_csv(tables / "model_comparison.csv", index=False)
```

Interpretation: features are lagged/trailing quantities; observations are not shuffled. The model selector uses validation MAE, and the final test period is held out.

Conclusion: forecasting results are experimental and do not imply future performance.
